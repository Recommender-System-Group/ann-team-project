```python
# ============================================
# 1. IMPORT DES LIBRAIRIES
# ============================================
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from tensorflow.keras.layers import Input, Embedding, Flatten, Dense, Concatenate, Dropout, BatchNormalization
from tensorflow.keras.models import Model
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt
import pickle

# ============================================
# 2. CHARGEMENT ET NETTOYAGE
# ============================================
df = pd.read_csv("goodreads_interactions.csv", nrows=50000)  #  

#  Filtrer les ratings à 0
df = df[df['rating'] > 0].copy()
df = df.reset_index(drop=True)

print(f"Lignes utilisées : {len(df)}")
print(f"Distribution :\n{df['rating'].value_counts().sort_index()}")

# ============================================
# 3. ENCODAGE
# ============================================
user_enc = LabelEncoder()
book_enc = LabelEncoder()

df['user'] = user_enc.fit_transform(df['user_id'])
df['book'] = book_enc.fit_transform(df['book_id'])

# ============================================
# 4. NORMALISATION
# ============================================
scaler = MinMaxScaler()
df[['is_read', 'is_reviewed']] = scaler.fit_transform(df[['is_read', 'is_reviewed']])

# ============================================
# 5. FEATURES ET TARGET
# ============================================
x_user  = df['user'].values.astype(np.int32)    #  numpy direct
x_book  = df['book'].values.astype(np.int32)    #  numpy direct
x_other = df[['is_read', 'is_reviewed']].values.astype(np.float32)  # numpy direct
y       = ((df['rating'] - 1) / 4.0).values.astype(np.float32)      #  numpy direct

print(f"\ny : min={y.min():.2f}  max={y.max():.2f}  moyenne={y.mean():.3f}")

# ============================================
# 6. SPLIT TRAIN / TEST
# ============================================
(x_user_train, x_user_test,
 x_book_train, x_book_test,
 x_other_train, x_other_test,
 y_train, y_test) = train_test_split(
    x_user, x_book, x_other, y,
    test_size=0.2,
    random_state=42
)

num_users = df['user'].nunique()
num_books = df['book'].nunique()

print(f"Users : {num_users}  |  Books : {num_books}")
print(f"Train : {len(y_train)}  |  Test : {len(y_test)}")

# ============================================
# 7. MODÈLE
# ============================================
user_input  = Input(shape=(1,), name='user_input')
book_input  = Input(shape=(1,), name='book_input')
other_input = Input(shape=(2,), name='other_input')

user_emb = Embedding(num_users, 32, name='user_embedding')(user_input) 
book_emb = Embedding(num_books, 32, name='book_embedding')(book_input)  

user_vec = Flatten()(user_emb)
book_vec = Flatten()(book_emb)

x = Concatenate()([user_vec, book_vec, other_input])
x = Dense(64, activation='relu')(x)
x = BatchNormalization()(x)
x = Dropout(0.3)(x)
x = Dense(32, activation='relu')(x)
x = Dropout(0.2)(x)

output = Dense(1, activation='sigmoid', name='output')(x) 

model = Model([user_input, book_input, other_input], output)
model.compile(optimizer='adam', loss='mse', metrics=['mae'])
model.summary()

# ============================================
# 8. ENTRAÎNEMENT
# ============================================
early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True,
    verbose=1
)

reduce_lr = ReduceLROnPlateau(
    monitor='val_loss',
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1
)

history = model.fit(
    [x_user_train, x_book_train, x_other_train],
    y_train,
    validation_split=0.2,
    epochs=20,       #  max 20
    batch_size=512,  #  grand batch
    callbacks=[early_stop, reduce_lr],
    verbose=1
)

# ============================================
# 9. ÉVALUATION
# ============================================
loss, mae = model.evaluate(
    [x_user_test, x_book_test, x_other_test],
    y_test, verbose=0
)

y_pred      = model.predict(
    [x_user_test, x_book_test, x_other_test],
    verbose=0, batch_size=512
).flatten()

y_pred_reel = np.clip(y_pred * 4 + 1, 1.0, 5.0)
y_test_reel = y_test * 4 + 1  # déjà numpy, pas besoin de .values

rmse      = np.sqrt(np.mean((y_pred_reel - y_test_reel) ** 2))
precision = np.mean(np.abs(y_pred_reel - y_test_reel) <= 1.0) * 100

print("\n" + "="*45)
print("RÉSULTATS FINAUX")
print("="*45)
print(f"MAE en étoiles : {mae * 4:.2f} ⭐")
print(f"RMSE           : {rmse:.3f} étoiles")
print(f"Précision ±1★  : {precision:.1f}%")
print(f"Prédictions    : min={y_pred_reel.min():.2f}  max={y_pred_reel.max():.2f}")
print("="*45)

# ============================================
# 10. COURBES D'APPRENTISSAGE
# ============================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.plot(history.history['loss'],     label='Train')
ax1.plot(history.history['val_loss'], label='Validation')
ax1.set_title('Loss (MSE)')
ax1.set_xlabel('Epoch')
ax1.legend()

ax2.plot(history.history['mae'],     label='Train')
ax2.plot(history.history['val_mae'], label='Validation')
ax2.set_title('MAE')
ax2.set_xlabel('Epoch')
ax2.legend()

plt.tight_layout()
plt.savefig('courbes_apprentissage.png')
plt.show()

# ============================================
# 11. EXEMPLE DE PRÉDICTION
# ============================================
sample_user  = x_user_test[0:1]   
sample_book  = x_book_test[0:1]
sample_other = x_other_test[0:1]

pred_norm   = model.predict([sample_user, sample_book, sample_other], verbose=0)
pred_rating = np.clip(pred_norm[0][0] * 4 + 1, 1.0, 5.0)

print(f"\nExemple de prédiction :")
print(f"  Rating réel   : {y_test[0] * 4 + 1:.1f} ⭐")
print(f"  Rating prédit : {pred_rating:.2f} ⭐")

# ============================================
# 12. SAUVEGARDE
# ============================================
model.save('book_recommendation_model.keras')

with open('user_encoder.pkl', 'wb') as f:
    pickle.dump(user_enc, f)
with open('book_encoder.pkl', 'wb') as f:
    pickle.dump(book_enc, f)
with open('scaler.pkl', 'wb') as f:
    pickle.dump(scaler, f)

print("\n Modèle et encodeurs sauvegardés !")

```

```python
# ============================================
# 13. RECOMMANDATIONS
# ============================================
def recommander_livres(user_idx, df, model, n=5):
    if user_idx not in df['user'].values:
        print(f" Utilisateur {user_idx} introuvable.")
        return []

    deja_lus      = set(df[df['user'] == user_idx]['book'].tolist())
    a_recommander = np.array(
        list(set(df['book'].unique()) - deja_lus), 
        dtype=np.int32   
    )

    if len(a_recommander) == 0:
        print(" L'utilisateur a tout lu.")
        return []

    users_array = np.full(len(a_recommander), user_idx, dtype=np.int32)
    other_array = np.zeros((len(a_recommander), 2), dtype=np.float32)

    preds = model.predict(
        [users_array, a_recommander, other_array],
        verbose=0, batch_size=2048
    ).flatten()

    preds_reelles = np.clip(preds * 4 + 1, 1.0, 5.0)
    top_idx       = preds_reelles.argsort()[-n:][::-1]

    return list(zip(a_recommander[top_idx].tolist(), preds_reelles[top_idx].tolist()))


# ── Test sur 3 utilisateurs actifs ────────────
users_actifs = (
    df.groupby('user')
    .filter(lambda x: len(x) >= 5)['user']
    .unique()[:3]
)

for user_idx in users_actifs:
    print(f"\n{'='*45}")
    print(f" User : {user_idx}  |  Livres notés : {len(df[df['user']==user_idx])}")
    recommandations = recommander_livres(user_idx, df, model, n=5)
    for rang, (livre, note) in enumerate(recommandations, 1):
        print(f"  #{rang}  Livre {livre:<8} → {note:.2f}  {'⭐'*round(note)}")
```
